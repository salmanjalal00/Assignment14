```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries are working!")
```

```python

# Load the Pakistani E-Commerce dataset

df = pd.read_csv("pakistani_ecommerce_dataset.csv")

print("Dataset loaded successfully!")
```

```python

# Display the first five rows

df.head()
```

```python

# Display the column names

print("Column Names:")
print(df.columns.tolist())

# Display the number of rows and columns

print("\nDataset Shape:")
print(df.shape)

# Display basic information

df.info()
```

```python

# Count the number of orders in each product category

category_counts = df['product_category'].value_counts()

print(category_counts)
```

```python

# Find the category with the highest frequency

most_popular = category_counts.idxmax()

print("Most Popular Product Category:", most_popular)
```

```python

# Calculate the average order value

average_order = df['order_value'].mean()

print("Average Order Value:", average_order)
```

```python

# Count orders by city

city_counts = df['city'].value_counts()

print(city_counts)
```

```python

# Find the city with the highest number of orders

top_city = city_counts.idxmax()

print("City with the highest number of orders:", top_city)
```

```python

# Count the use of each payment method

payment_counts = df['payment_method'].value_counts()

print(payment_counts)
```

```python

# Identify the most popular payment method

popular_payment = payment_counts.idxmax()

print("Most Used Payment Method:", popular_payment)
```

```python

# Calculate average order value for each city

average_by_city = df.groupby('city')['order_value'].mean()

print(average_by_city)
```

```python

# Draw the bar chart

average_by_city.plot(kind='bar', figsize=(9, 5))

plt.title('Average Order Value by City')
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Load Student Lifestyle dataset

df2 = pd.read_csv("student_lifestyle_dataset.csv")

print("Student dataset loaded successfully!")
```

```python

# Display first five rows

df2.head()
```

```python

# Display column names and dataset information

print(df2.columns.tolist())

print(df2.shape)

df2.info()
```

```python

# Calculate average study hours

average_study = df2['study_hours'].mean()

print("Average Study Hours:", average_study)
```

```python

# Compare average sleep hours by gender

sleep_comparison = df2.groupby('gender')['sleep_hours'].mean()

print(sleep_comparison)
```

```python

# Find the gender with the highest average sleep

more_sleep = sleep_comparison.idxmax()

print("Gender that sleeps more on average:", more_sleep)
```

```python

# Calculate correlation between screen time and sleep

correlation = df2['screen_time'].corr(df2['sleep_hours'])

print("Correlation:", correlation)
```

```python

# Visualize the relationship

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df2, x='screen_time', y='sleep_hours')

plt.title('Relationship Between Screen Time and Sleep Hours')
plt.xlabel('Screen Time')
plt.ylabel('Sleep Hours')

plt.show()
```

```python

# Draw histogram of daily transport cost

plt.figure(figsize=(8, 5))

plt.hist(df2['daily_transport_cost'], bins=10, edgecolor='black')

plt.title('Distribution of Daily Transport Cost')
plt.xlabel('Daily Transport Cost')
plt.ylabel('Number of Students')

plt.show()
```

```python

# Calculate average internet usage by gender

internet_comparison = df2.groupby('gender')['internet_usage'].mean()

print(internet_comparison)
```

```python

# Create bar chart

internet_comparison.plot(kind='bar', figsize=(7, 5))

plt.title('Average Internet Usage by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Internet Usage')

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
